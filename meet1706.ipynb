```python
from bs4 import BeautifulSoup
import pandas as pd

website = "30 Most Popular Movies Right Now_ What to Watch In Theaters and Streaming __ Rotten Tomatoes – Movie and TV News.html"

file = open(website, encoding = "utf-8")

res = file.read()
# print(res)

soup = BeautifulSoup(res, "html.parser")
# print(soup)

movie_title = []
ratings = []
year = []

movies = soup.find_all('div', class_="row countdown-item-title-bar")
for movie in movies:
    movie_title.append(movie.h2.a.text)
    ratings.append(movie.find('span', class_="tMeterScore").text)

for movie_year in movies:
    year.append(movie_year.find('span', class_="subtle start-year").text)
year

df = pd.DataFrame({
    'Movie_Titles' : movie_title,
    'Ratings' : ratings,
    'Years' : year
})
df
# df.isnull().sum()
```

```python
from bs4 import BeautifulSoup
import pandas as pd

website = "Fact-checks _ PolitiFact.html"

file = open(website, encoding = "utf-8")

res = file.read()
# print(res)

soup = BeautifulSoup(res, "html.parser")
# print(soup)

statements =[]
sources = []
dates = []

statement = soup.find_all('article', class_="m-statement")
for state in statement:
    statements.append(state.find('div', class_="m-statement__quote").text.strip())
statements

source = soup.find_all('article', class_="m-statement")
for state in source:
    sources.append(state.find('a').text.strip())
sources

date = soup.find_all('div', class_="m-statement__meta")
for state in date:
    # date_on = state.div.text.strip().split('stated on')[1]
    date_on = state.div.text.strip()[9:]
    dates.append(date_on.split('in')[0])
dates

df = pd.DataFrame({
    'Statements' : statements,
    'Names' : sources,
    'Dates' : dates
})
df
```

```python
from bs4 import BeautifulSoup
import pandas as pd

website = "Certified used Mercedes-Benz for sale.html"

file = open(website, encoding = "utf-8")
res = file.read()
# print(res)

soup = BeautifulSoup(res, "html.parser")
# print(soup)

names = []
mileages = []
prices = []
review_Count = []
dealer_names = []
ratings = []

car_name = soup.find_all('a', class_="vehicle-card-link")
for c_name in car_name:
    names.append(c_name.find('h2', class_="title").text)
names

car_mileage = soup.find_all('div', class_="mileage")
for c_mileage in car_mileage:
    mileage.append(c_mileage.text.strip())
mileages = mileage.remove('Mileage')

car_price = soup.find_all('div', class_="price-section")
for c_price in car_price:
    prices.append(c_price.span.text.strip())
prices

car_review_count = soup.find_all('div', class_="vehicle-dealer")
for c_rcount in car_review_count:
    review_Count.append(c_rcount.find('span', class_="sds-rating__link").text.strip())
    dealer_names.append(c_rcount.find('div', class_="dealer-name").text.strip())
    
review_Count
dealer_names

car_ratings = soup.find_all('div', class_="sds-rating")
for c_ratings in car_ratings:
    ratings.append(c_ratings.find('span', class_="sds-rating__count").text.strip())
ratings


df = pd.DataFrame({
    'Names' : names,
    'Mileage' : mileages,
    'Prices' : prices,
    'Review_Count' : review_Count,
    'Dealer_names' : dealer_names,
    'Ratings' : ratings
})
df
```
